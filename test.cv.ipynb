```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Generate dummy data
np.random.seed(42)
n_weeks = 5 * 52
trend = np.linspace(0, 50, n_weeks)
seasonal = 10 * np.sin(2 * np.pi * np.arange(n_weeks) / 52)
noise = np.random.normal(0, 2, n_weeks)
data = trend + seasonal + noise
dates = pd.date_range(start='2018-01-01', periods=n_weeks, freq='W')
dummy_data = pd.Series(data, index=dates)

# Plot the generated data
dummy_data.plot(title="Dummy Data with Trend and Seasonality", figsize=(10, 6))
plt.show()

# Fit SARIMAX model and predict
initial_train_weeks = 3 * 52
model = SARIMAX(dummy_data[:initial_train_weeks], order=(1, 1, 1), seasonal_order=(1, 1, 1, 52))
results = model.fit()

forecast_steps = 4

# Iterate and update model
predictions = []
for i in range(8):
    # print iteration
    import datetime
    print(datetime.datetime.now(), f"Iteration {i+1}")
    # Predict 4 weeks forward
    pred = results.get_forecast(steps=forecast_steps)
    predictions.extend(pred.predicted_mean)
    
    # Update model with 1 additional week
    next_week = initial_train_weeks + forecast_steps * (i + 1)
    print('next week', next_week)
    if next_week + 4 > len(dummy_data):
        break
    model = SARIMAX(dummy_data[:next_week], order=(1, 1, 1), seasonal_order=(1, 1, 1, 52))
    results = model.fit(disp=False)


```

```python
# Convert predictions to a Pandas Series
pred_dates = pd.date_range(start=dummy_data.index[initial_train_weeks], periods=12 * 8, freq='W')
predictions_series = pd.Series(predictions, index=pred_dates)

print(predictions)

# Plot the predictions
plt.figure(figsize=(10, 6))
plt.plot(dummy_data, label='Actual Data')
plt.plot(predictions_series, label='Predicted Data', linestyle='--')
plt.title("SARIMAX Predictions")
plt.legend()
plt.show()

# calculate rmspe and rmse and print
from sklearn.metrics import mean_squared_error
from math import sqrt
def rmspe(y_true, y_pred):
    return np.sqrt(np.mean(np.square((y_true - y_pred) / y_true)))

def rmse(y_true, y_pred):
    return sqrt(mean_squared_error(y_true, y_pred))

test_actual = dummy_data[initial_train_weeks: initial_train_weeks + predictions_series.shape[0]]
print(test_actual.shape)
print(predictions_series.shape)

print('RMSPE:', rmspe(test_actual, predictions_series))
print('RMSE:', rmse(test_actual, predictions_series))
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import TimeSeriesSplit
import matplotlib.pyplot as plt

# Create Dummy Time Series Data (5 Years)
np.random.seed(42)  # for reproducibility
index = pd.date_range(start='2015-01-01', periods=365*5, freq='D')
trend = 0.01 * np.arange(len(index))
seasonality = 2 * np.sin(2 * np.pi * index.dayofyear / 365)
noise = np.random.randn(len(index))
data = pd.Series(trend + seasonality + noise, index=index)

# Cross-Validation Setup
tscv = TimeSeriesSplit(n_splits=4, test_size=365, gap=0)  # 4 folds, 1-year test, no gap

# Iteration and Visualization
for i, (train_index, test_index) in enumerate(tscv.split(data)):
    print(f"\nFold {i + 1}:")
    print("  Train:", data.index[train_index[0]], "-", data.index[train_index[-1]])
    print("  Test:", data.index[test_index[0]], "-", data.index[test_index[-1]])

    # Visualize the split
    plt.figure(figsize=(12, 4))
    plt.plot(data[train_index], label='Train', color='blue')
    plt.plot(data[test_index], label='Test', color='orange')
    plt.xlabel('Date')
    plt.ylabel('Value')
    plt.title(f'Time Series Cross Validation - Fold {i+1}')
    plt.legend()
    plt.grid(axis='y')
    plt.tight_layout()
    plt.show()
```

```python
# time series cross validation
# use dummy data with some trend and seasonality of 5 years
# use 3 years of data for training and 1 year for testing
# have 1 month gap between training and testing data
# keep size of training data constant and move the testing data

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import TimeSeriesSplit

# generate dummy data
np.random.seed(0)
n = 60
x = np.arange(n)
y = 0.5 * x + 2 + np.random.normal(0, 1, n)
y[12:24] += 5
y[36:48] += 5

# plot dummy data
plt.plot(x, y)
plt.show()

# time series cross validation
tscv = TimeSeriesSplit(n_splits=3)

# train and test

for train_index, test_index in tscv.split(x):
    print("TRAIN:", train_index, "TEST:", test_index)
    X_train, X_test = x[train_index], x[test_index]
    y_train, y_test = y[train_index], y[test_index]
    print(X_train, X_test)
    print(y_train, y_test)
    print()

    # train model
    model = LinearRegression()
    model.fit(X_train.reshape(-1, 1), y_train)

    # predict
    y_pred = model.predict(X_test.reshape(-1, 1))

    # evaluate
    mse = mean_squared_error(y_test, y_pred)
    print("MSE:", mse)

    # plot
    plt.plot(X_train, y_train, label="train")
    plt.plot(X_test, y_test, label="test")
    plt.plot(X_test, y_pred, label="pred")
    plt.legend()
    plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Dummy Data Generation
np.random.seed(42)  # for reproducibility
years = pd.date_range(start='2015-01-01', end='2019-12-31', freq='D')
trend = 0.1 * np.arange(len(years))  # Upward linear trend
seasonality = 20 * np.sin(2 * np.pi * years.dayofyear / 365)  # Annual cycle
noise = 5 * np.random.randn(len(years))
data = pd.Series(trend + seasonality + noise, index=years)

# 2. Cross-Validation Loop
train_period = 3 * 365  # 3 years
test_period = 365      # 1 year
gap_period = 30        # 1 month

results = []
for i in range(0, len(data) - train_period - gap_period - test_period, 10):
    train_start = data.index[i]
    train_end = data.index[i + train_period - 1]
    test_start = data.index[i + train_period + gap_period]
    test_end = data.index[i + train_period + gap_period + test_period - 1]

    # print range of train and test
    print(train_start, '==>', train_end, '||', test_start,'==>', test_end)
    # render the range of train and test
    plt.plot(data.loc[train_start:train_end])
    plt.plot(data.loc[test_start:test_end])
    plt.show()


    train_data = data[train_start:train_end]
    test_data = data[test_start:test_end]

    # 3. Simple Model (Example)
    X_train = np.arange(len(train_data)).reshape(-1, 1)
    model = LinearRegression().fit(X_train, train_data)
    X_test = np.arange(len(train_data), len(train_data) + len(test_data)).reshape(-1, 1)
    predictions = model.predict(X_test)

    # 4. Evaluation (Example)
    mse = np.mean((predictions - test_data)**2)
    results.append({'train_start': train_start, 'mse': mse})

# 5. Analyze Results
results_df = pd.DataFrame(results)
print(results_df)  # See how the error varies over time
```
