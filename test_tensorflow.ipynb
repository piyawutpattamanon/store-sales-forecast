```python
import tensorflow as tf
import numpy as np

# Generate dummy dataset
np.random.seed(42)
X = np.random.rand(100, 2)
true_weights = np.array([3, 5])
true_bias = 10
y = np.dot(X, true_weights) + true_bias + np.random.randn(100) * 0.1

# Convert to TensorFlow tensors
X = tf.constant(X, dtype=tf.float32)
y = tf.constant(y, dtype=tf.float32)

# Initialize model parameters
W = tf.Variable(tf.random.normal([2]), name='weights')
b = tf.Variable(tf.random.normal([1]), name='bias')

# Define the learning rate and number of epochs
learning_rate = 0.1
epochs = 1000

# Define the linear regression model
def linear_regression(X):
    return tf.reduce_sum(W * X, axis=1) + b

# Define the mean squared error loss function
def mean_squared_error(y_true, y_pred):
    return tf.reduce_mean(tf.square(y_true - y_pred))

# Training loop
for epoch in range(epochs):
    with tf.GradientTape() as tape:
        y_pred = linear_regression(X)
        loss = mean_squared_error(y, y_pred)

    # Compute gradients
    gradients = tape.gradient(loss, [W, b])

    # Update parameters using gradient descent
    W.assign_sub(learning_rate * gradients[0])
    b.assign_sub(learning_rate * gradients[1])

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss.numpy()}, Weights: {W.numpy()}, Bias: {b.numpy()}")

# Final parameters
print(f"Final Weights: {W.numpy()}, Final Bias: {b.numpy()}")
```
