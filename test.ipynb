```python
import csv
import tqdm

rows = []
with open('train.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in tqdm.tqdm(reader):
        rows.append(row)
```

```python
import polars as pl

df = pl.DataFrame(rows[:])
df
```
